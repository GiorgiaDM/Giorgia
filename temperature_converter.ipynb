```python
while True:
    unit = input("Convert to Celsius or Fahrenheit? ").strip().lower()

    if unit == "fahrenheit":
        temperature = float(input("Enter temperature in Celsius: "))
        fahrenheit = (temperature * 9/5) + 32
        print(f"{temperature} Celsius is equal to {fahrenheit} Fahrenheit")

    elif unit == "celsius":
        temperature = float(input("Enter temperature in Fahrenheit: "))
        celsius = (temperature - 32) * 5/9
        print(f"{temperature} Fahrenheit is equal to {celsius} Celsius")

    else:
        print("Invalid choice, please enter Celsius or Fahrenheit.")

    again = input("Do you want to convert another temperature? (yes/no): ").strip().lower()

    if again != "yes":
        break
```
